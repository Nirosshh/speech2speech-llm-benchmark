```python
from huggingface_hub import InferenceClient

model_id = "meta-llama/Llama-3.1-8B"  # base model (you have access)
#model_id = "meta-llama/Llama-3.1-8B-Instruct"

client = InferenceClient(model=model_id, token="HF_REDACTED")
```

```python
def llama_chat_prompt(user_msg, system_msg="You are a helpful assistant."):
    return (
        "<|begin_of_text|>"
        "<|start_header_id|>system<|end_header_id|>\n\n"
        f"{system_msg}<|eot_id|>"
        "<|start_header_id|>user<|end_header_id|>\n\n"
        f"{user_msg}<|eot_id|>"
        "<|start_header_id|>assistant<|end_header_id|>\n\n"
    )

prompt = llama_chat_prompt("Give me three bullet points about the Apriori algorithm.")
```

```python
text = client.text_generation(
    prompt=prompt,
    max_new_tokens=200,
    temperature=0.7,
    top_p=0.9,
    repetition_penalty=1.05,
    do_sample=True,
    return_full_text=False,
)
print(text)
```
