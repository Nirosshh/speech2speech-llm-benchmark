```python
from huggingface_hub import InferenceClient

HF_TOKEN = "HF_REDACTED"
MODEL_ID = "meta-llama/Llama-3.1-8B-Instruct"

def llama_chat_prompt(user_msg, system_msg="You are a helpful assistant."):
    return (
        "<|begin_of_text|>"
        "<|start_header_id|>system<|end_header_id|>\n\n"
        f"{system_msg}<|eot_id|>"
        "<|start_header_id|>user<|end_header_id|>\n\n"
        f"{user_msg}<|eot_id|>"
        "<|start_header_id|>assistant<|end_header_id|>\n\n"
    )

prompt = llama_chat_prompt("Explain cross-validation in 2 short sentences.")

client = InferenceClient(model=MODEL_ID, token=HF_TOKEN, provider="hf-inference")
text = client.text_generation(
    prompt=prompt,
    max_new_tokens=200,
    temperature=0.7,
    top_p=0.9,
    repetition_penalty=1.05,
    do_sample=True,
    return_full_text=False,
)
print(text)
```
