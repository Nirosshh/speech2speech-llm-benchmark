```python
model = genai.GenerativeModel("gemini-1.5-flash")  # or "gemini-1.5-pro"
```

```python
prompt = "Hi"
```

```python
prompt = "why latency of the reslponse vary each time?"
```

```python
# --- Start request & measure TTFT/TTFX ---
ttft = None
start = time.time()              # request start (for TTFT)
resp = model.generate_content(
    prompt,
    stream=True,
    safety_settings={HarmCategory.HARM_CATEGORY_HATE_SPEECH: HarmBlockThreshold.BLOCK_NONE}
)

full = []
for chunk in resp:
    text = chunk.text or ""
    if not text:
        continue
    if ttft is None:
        ttft = time.time() - start
        print(f"\nTTFT: {ttft:.3f}s\n")
    full.append(text)
    print(text, end="", flush=True)

# finished streaming
ttfx = time.time() - start
full_text = "".join(full)

print("\n\n" + "-"*50)
print(f"TTFT (first token): {ttft:.3f}s")
print(f"TTFX (final token): {ttfx:.3f}s")
print(f"Output chars: {len(full_text)}")
```
